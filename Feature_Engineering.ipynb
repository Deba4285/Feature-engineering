1.  What is a parameter?    
    ans :- A parameter is a value inside a machine learning model that is learned from the training data. Parameters help the model make predictions by capturing patterns in the data. For example, in linear regression, the slope and intercept are parameters. During training, the model adjusts these parameters to minimize errors and improve prediction accuracy.

2. What is correlation?

    ans :- Correlation is a statistical measure that describes the relationship between two variables. It shows how strongly and in what direction one variable changes when another variable changes. Correlation values range from -1 to +1. A value close to +1 indicates a strong positive relationship, while a value close to -1 indicates a strong negative relationship.

3. What Does Negative Correlation Mean?

     ans : - Negative correlation means that when one variable increases, the other variable decreases. The correlation coefficient is less than zero. For example, the relationship between product price and demand is often negatively correlated because higher prices usually lead to lower demand.

4. How Does Loss Value Help in Determining Whether the Model is Good or Not?

   ans : -Loss value measures the difference between the model's predicted output and the actual output. A lower loss value indicates that the model's predictions are closer to the true values, suggesting better performance. During training, the objective is to minimize the loss value. However, both training and testing loss should be considered to avoid overfitting.

5. What Are Continuous and Categorical Variables?

   ans : - Continuous variables are numerical variables that can take any value within a range, such as height, weight, and temperature. Categorical variables represent groups or categories, such as gender, color, and city names. Continuous variables are used for mathematical calculations, while categorical variables need encoding before being used in most machine learning models.

6. How Do We Handle Categorical Variables in Machine Learning? What Are the Common Techniques?

   ans : - Categorical variables are converted into numerical form because machine learning algorithms work with numbers. Common techniques include Label Encoding, which assigns a unique number to each category, and One-Hot Encoding, which creates separate binary columns for each category. Other techniques include Ordinal Encoding, Target Encoding, and Frequency Encoding.

7. What Do You Mean by Training and Testing a Dataset?

     ans : - Training is the process of teaching a machine learning model using historical data so that it can learn patterns and relationships. Testing is the process of evaluating the trained model on unseen data to measure its performance. This helps determine how well the model can generalize to new data.

8. What is sklearn.preprocessing?

    ans : -sklearn.preprocessing is a module in the Scikit-learn library used for data preprocessing. It provides tools for scaling, normalizing, encoding categorical variables, and transforming data before training machine learning models. Proper preprocessing improves model performance and training efficiency.

9. What is a Test Set?

    ans: -  test set is a portion of the dataset reserved for evaluating a trained machine learning model. It contains unseen data that was not used during training. The test set helps determine how well the model performs on new data and provides an unbiased estimate of model accuracy.

10. How Do We Split Data for Model Fitting (Training and Testing) in Python?

    ans : - In Python, data is commonly split using the train_test_split() function from Scikit-learn. This function divides the dataset into training and testing subsets according to a specified ratio. A common split is 80% training data and 20% testing data. This ensures proper evaluation of the model.

11. How Do You Approach a Machine Learning Problem?

    ans : - A machine learning problem is approached by understanding the business objective, collecting data, performing data cleaning and preprocessing, conducting exploratory data analysis (EDA), selecting relevant features, choosing an appropriate algorithm, training the model, evaluating performance, tuning hyperparameters, and finally deploying the model for real-world use.

12. . Why Do We Have to Perform EDA Before Fitting a Model to the Data?

Exploratory Data Analysis (EDA) helps understand the structure, quality, and characteristics of the data. It identifies missing values, outliers, distributions, and relationships between variables. EDA enables better feature selection and preprocessing decisions, leading to improved model performance and more reliable predictions.

13. What is Correlation?

    ans : - Correlation is a measure of how two variables move in relation to each other. It indicates both the strength and direction of the relationship. Positive correlation means both variables move in the same direction, while negative correlation means they move in opposite directions.

14. What Does Negative Correlation Mean?

    ans : - Negative correlation occurs when an increase in one variable is associated with a decrease in another variable. The correlation coefficient lies between 0 and -1. For example, as fuel prices increase, vehicle usage may decrease, showing a negative correlation.

15. How Can You Find Correlation Between Variables in Python?

    ans : - Correlation can be calculated in Python using Pandas with the corr() function. This function generates a correlation matrix showing relationships among numerical variables. It helps identify strongly related features that may be useful or redundant in machine learning models.

16. What is Causation? Explain Difference Between Correlation and Causation with an Example.

    ans : - Causation means that one event directly causes another event to occur. Correlation only indicates that two variables move together but does not prove a cause-and-effect relationship. For example, ice cream sales and drowning incidents may increase together during summer. They are correlated, but ice cream sales do not cause drowning; hot weather influences both.

17. What is an Optimizer? What Are Different Types of Optimizers? Explain Each with an Example.

    ans : - An optimizer is an algorithm used to update model parameters and minimize the loss function during training. Common optimizers include Gradient Descent, which updates parameters using the entire dataset; Stochastic Gradient Descent (SGD), which updates using one sample at a time; RMSProp, which adapts learning rates for each parameter; and Adam, which combines momentum and adaptive learning rates. Adam is widely used because it converges quickly and efficiently.

18. What is sklearn.linear_model?

    ans : - sklearn.linear_model is a module in Scikit-learn that contains linear machine learning algorithms such as Linear Regression, Logistic Regression, Ridge Regression, and Lasso Regression. These models are used for prediction and classification tasks based on linear relationships between features and target variables.

19. What Does model.fit() Do? What Arguments Must Be Given?

    ans : - The model.fit() method trains a machine learning model using the provided data. It learns patterns by adjusting model parameters. The required arguments are feature data (X) and target values (y). After training, the model is ready to make predictions on new data.

20. What Does model.predict() Do? What Arguments Must Be Given?

    ans : - The model.predict() method uses a trained model to generate predictions for new input data. The required argument is the feature dataset (X) containing observations for which predictions are needed. The method returns the predicted output values.
